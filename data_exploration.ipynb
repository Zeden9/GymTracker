```python

normalized_data = pd.read_csv("data/processed/normalized_notes.csv")
normalized_data
normalized_data["Date"] = pd.to_datetime(normalized_data["Date"])
type(normalized_data["Date"].iloc[0])
```

```python
df = normalized_data
df['Date'] = pd.to_datetime(df['Date'])
df['Volume'] = df['Weight'] * df['Reps']
volume_df = df.groupby(['Date', 'Exercise'])['Volume'].sum().reset_index()

wybrane_cwiczenie = "ez bar curl" 

filtered_data = volume_df[volume_df['Exercise'] == wybrane_cwiczenie]


plt.figure(figsize=(10, 5))
sns.lineplot(data=filtered_data, x='Date', y='Volume', marker='o', color='royalblue', linewidth=2.5)

# Dodanie opisu i stylizacja
plt.title(f'Postęp Objętości: {wybrane_cwiczenie.upper()}', fontsize=14, pad=20)
plt.xlabel('Data treningu')
plt.ylabel('Całkowita Objętość (kg)')
plt.grid(True, linestyle=':', alpha=0.7)


plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
